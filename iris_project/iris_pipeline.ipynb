# Iris ML pipeline

This notebook documents the local Mage-style Iris pipeline. The block graph is defined in `metadata.yaml` and the implementation is split into loaders, transformers, and exporters.

Flow: `load_iris -> validate_iris -> prepare_data -> train_model -> save_results/evaluate_model -> export_metrics`.

```python
from pathlib import Path
from iris_project.loaders.load_iris import load_data
from iris_project.run_pipeline import run
from iris_project.transformers.validate_iris import test as test_validation
from iris_project.transformers.validate_iris import transform as validate_iris

validated = validate_iris(load_data())
test_validation(validated)
output_dir = Path('iris_project/notebook-output')
accuracy = run(max_depth=3, output_dir=str(output_dir), test=True)
print(f'test_accuracy={accuracy:.4f}')
```
